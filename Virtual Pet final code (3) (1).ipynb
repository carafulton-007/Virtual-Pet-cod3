```python
import random
import time
print("Welcome to Virtual Pet.Python!!! I hope you enjoy my game!!")

#Scene 1

time.sleep(1.5)
print('Shopkeeper: "Oh hello!! Who are you?"')
time.sleep(0.5)
name = input("Enter name: ")
time.sleep(1.5)
print(f'Shopkeeper: "Welcome {name} to my pet shop! Would you like an egg?"')
time.sleep(0.5)
answer = input("Enter yes or no: ")

if answer.lower()== 'yes':
    time.sleep(1.5)
    print('Shopkeeper: "Yes? Ah well you are just in luck for I have an egg that I am willing to give you for free. Courtesy of myself."')
else:
    time.sleep(1.5)
    print('Shopkeeper: "Nonsense!!!! You shall have one anyway, free of charge!"')

time.sleep(0.5)
print('system : shopkeeper hands you the egg.')
time.sleep(0.5)
hatch1 = input('Type anything to hatch your egg!: ')
print('system: Hatching.....')
pets=('cat', 'dog', 'bird', 'snake', 'hamster', 'sheep', 'chicken', 'lion',  'lizard')
hatched_pet = random.choice(pets)
time.sleep(3.5)
chosen = print(f'Congratulations!! You Hatched a {hatched_pet}!!')
time.sleep(0.5)
pet_name= input("Name your pet: ")
time.sleep(0.5)
print(f'system: it is getting late so you and your newfound pet {pet_name} head back home')
time.sleep(3)
#scene 2

print('system: Scene 2:')
time.sleep(0.5)
print(f'system : You and your pet {pet_name} have arrived home.')
time.sleep(2)
print('system : In order to keep your pet alive and well your pet must have completed basic tasks such as eating,sleeping,playing and kept healthy!!')
time.sleep(1.5)
print("system: Doing this will keep your pet's happiness high (below 50 causes your pet to die.)")
time.sleep(3)
print('system : In a second you will be promted to type any command. This will appear after every action. ')
print("system : If you type 'stats', you'll be able to view your pet's hunger, energy, affection and overall happiness rate (average of other stats).")
time.sleep(2.5)
print("system : If you type 'game', you can enter a mini-game where you have to solve basic equations to increase Pet Stats.")
time.sleep(3.5)
print("system : Failing ANY question will result in a decrease of stats so make sure you get the right answer.")
time.sleep(0.5)


#stats


hunger=100
affection=100
energy=100
health=100
happy= (hunger+affection+energy)/3

count=0

# mini game 
 
def mini_game():
    global hunger, affection, energy, health, happy
    correct = 0
    wrong=0
    
    for i in range(5):
        num1 = random.randint(1, 20)
        num2 = random.randint(1, 30)
        
        correct_answer = num1 + num2
        

        print(f" Game: What is {i + 1}: {num1} + {num2} = ?")
        user_answer = int(input("Your answer: "))
        

        if user_answer == correct_answer:
            print("Correct!")
            correct+=1 
        else:
            print("Incorrect!!")
            wrong+=1
      
    
    if wrong>0:
        hunger-=4
        health-=10
        energy-=5
        affection-=3
    elif correct>0:  
        hunger+=4
        health+=10
        energy+=5
        affection+=3
    else:
        wrong=wrong
        correct=correct
    
    text2 = input("system: Type desired command: ")



# commands


while True:

    count += 1

    if count % 20 == 0:
        hunger -= 2
    if count % 30 == 0:
        affection -= 2
    if count % 60 == 0:
        energy -= 2
    if count % 35 == 0:
        health -=5


    command = input("\nsystem: Type A Command (Stats/game/quit): ").lower()

    if command.lower()== "stats":
        happy = (hunger + energy + affection) / 3
        print(f"Hunger: {hunger}")
        print(f"Energy: {energy}")
        print(f"Affection: {affection}")
        print(f"Health: {health}")
        print(f"Happiness: {happy}")

    elif command.lower()== "game":
        mini_game()

    elif command.lower()== "quit":
        print("Thanks for playing!")
        break

    else:
        print("Unknown command.")


    if happy<50:
        print(f'system: {pet_name} has died!!! You should have taken better care of them!!')
        break
    else:
        happy=happy
```
